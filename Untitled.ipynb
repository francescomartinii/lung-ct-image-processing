```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image1.jpg", 0)

plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
_, thresh = cv2.threshold(img, 120, 255, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Segmented")
plt.show()
```

```python
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(thresh, cmap='gray')
plt.title("Segmented")

plt.show()
```

```python
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(thresh, cmap='gray')
plt.title("Segmented")
plt.axis('off')

plt.show()
```

```python
img_eq = cv2.equalizeHist(img)

plt.imshow(img_eq, cmap='gray')
plt.title("Enhanced Image")
plt.axis('off')
plt.show()
```

```python
edges = cv2.Canny(img, 50, 150)

plt.imshow(edges, cmap='gray')
plt.title("Edges")
plt.axis('off')
plt.show()
```
